mediapipe is developed by google. It is a framework just like pytesseract or opencv. It is an opensource & it is everything related to computer vision i.e. OCR, Object Detection, Classification, Face Detection, human Pose detection or applying filter on instagram, changing colour of hair, box tracking, motion tracking, etc.

Mediapipe FaceMesh is a solution that estimates 468 facemarks or facelandmarks in realtime even on Mobile devices. It employs machine learning to infer 3D facial surface requiring only a single camera input without the need for a dedicated Depth Sensor. Its better to use GPU or TPU to deliver realtime performance critical for live experiences.

Similarly in mediapipe hands 21 hand knuckle coordinates are there.

```python
import cv2
import mediapipe as mp
import numpy as np
from google.colab.patches import cv2_imshow
```

```python
mp_hands=mp.solutions.hands  # It loads Mediapipes hand tracking module
mp_hands
```

```python
hands=mp_hands.Hands(static_image_mode=True,max_num_hands=2,min_detection_confidence=0.5)
# it creates a hand detection model with parameters static_image_mode = True, when its true it processes each
# image independently and it is better for images. When it is False, it uses tracking to improve hand detection in
# a video
```

```python
image=cv2.imread("/content/OIP.jpg")
rgb_image=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
results = hands.process(rgb_image)
```

```python
results
```

```python
mp_draw = mp.solutions.drawing_utils # Loads drawing utilities to visualise keypoints
if results.multi_hand_landmarks:# if any hand detected in results, then draw dots and lines connecting key hand points
  for hand_landmarks in results.multi_hand_landmarks:
    mp_draw.draw_landmarks(image,hand_landmarks,mp_hands.HAND_CONNECTIONS)
```

```python
cv2_imshow(image)
```

```python
mp_face_mash = mp.solutions.face_mesh
face_mesh=mp_face_mash.FaceMesh(static_image_mode=True,max_num_faces=2,min_detection_confidence=0.5)
```

```python
from google.colab import files
upload=files.upload()
img_path=list(upload.keys())[0]

image=cv2.imread(img_path)
rgb_image=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
results=face_mesh.process(rgb_image)  # image is getting processed with the help of face_mesh

if results.multi_face_landmarks:
  for face_landmarks in results.multi_face_landmarks:
    mp_draw.draw_landmarks(image,face_landmarks,mp_face_mash.FACEMESH_CONTOURS)
cv2_imshow(image)
```
