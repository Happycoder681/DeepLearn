```python
import mediapipe as mp
import cv2
import time
```

```python
mp_holistic=mp.solutions.holistic
holistic_model=mp_holistic.Holistic()
mp_drawing=mp.solutions.drawing_utils
```

```python
import sys
sys.executable
```

```python
# !{sys.executable} -m pip install mediapipe==0.10.13              #This is for the time when the 1st code doen't Run
```

```python
# capture the video
capture=cv2.VideoCapture(0)
# for frames per second calculation 
previousTime=0
currentTime=0
while capture.isOpened():
    # capture the frame
    ret,frame=capture.read()
    #resizing frame
    frame=cv2.resize(frame,(800,600))
    # convert BGR to RGB
    image=cv2.cvtColor(frame,cv2.COLOR_BGR2RGB)
    #detect landmarks
    image.flags.writeable=False
    results=holistic_model.process(image)
    image.flags.writeable=True
    
    #convert back image to BGR
    image=cv2.cvtColor(image,cv2.COLOR_RGB2BGR)

    # facial landmarks
    mp_drawing.draw_landmarks(
        image, results.face_landmarks,
        mp_holistic.FACEMESH_CONTOURS,
        mp_drawing.DrawingSpec( color=(255,0,255),
                               thickness=1, circle_radius=1),
        mp_drawing.DrawingSpec( color=(255,0,255),
                               thickness=1, circle_radius=1))
    # Right Hand landmarks
    mp_drawing.draw_landmarks(
        image,results.right_hand_landmarks,mp_holistic.HAND_CONNECTIONS)
    # Left hand landmarks
    mp_drawing.draw_landmarks(
        image,results.left_hand_landmarks,mp_holistic.HAND_CONNECTIONS)

    # calculating FPS on image
    currentTime=time.time()
    fps=1/(currentTime-previousTime)
    previousTime=currentTime
    # Displaying FPS on image
    cv2.putText(image, str(int(fps))+"FPS",(10,70),cv2.FONT_HERSHEY_COMPLEX,1,(0,255,0),2)

    # show the frame
    cv2.imshow("Media Pipe Practical", image)
    key=cv2.waitKey(5)
    if key==ord('q'):
        break
capture.release()
cv2.destroyAllWindows()
```
