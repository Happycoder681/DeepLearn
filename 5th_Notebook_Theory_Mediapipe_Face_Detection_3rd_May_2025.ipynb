RCNN i.e Region Based convolutional neural network is a type of Deep Learning Model used for Object Detection. It combines the power of CNNs with Region proposals to detect objects in images. It is trained on coco dataset (Common Objects in Context).

**Its working**

Region proposals - Selective search or other region proposal techniques is used to generate regions of interest in an image. these are the areas where objects are lightly to be located.
**2nd Step**
CNN features extraction - each proposed region is passed through a CNN to extract features. eg. Edge, texture, etc.
Classification - A classifier like SVM is used to classify the regions as containing an object. eg. cat, car, etc
Bounding Box Refinement - The bounding box says around the detected objects are refined using a method like regression to improve the accuracy of object localisation

RCNN variants
1. Fast RCNN - it is faster than original RCNN because it processes the entire image once and then applies ROI cooling for each proposed region
2. Faster RCNN does further improvement with the introduction of a region proposal network that generate region proposals end to end making it faster and more efficient
3. Masked RCNN - an extension of faster RCNN that also outputs object mask i.e segmentation and not just the bounding box. The RCNN output is like [class label,bounding box coordinates,confidence score]
Otherwise to get the class score softmax activation function is used.


SSD Single Shot MultiBox Detector for Object Detection:
These are developed to detect object in real time. Mobilenet SSD is specially developed for working on mobiles. here
1. the feature extraction is applied using CNN and then the 6 layers of convolutional layer is applied. It predicts around 8000 bounding boxes on image and then select the best box. this is also called as Non-Max suppresion.
In Non-Max Suppresion NMS, we apply intersection over union and then select the box which has maximum overlap. The best box is the one with the highest value of IOU i.e 1 and lowest 0.
YOLO You Only Look Once - YOLO is extremely fast multi object detection algorithm that uses CNN to detect and identify objects. it divies the onput image into a grid that predicts a certain number of bounding boxes and the class probability. here each grid cell is responsible for predicting an object if the object centre falls within it. It applies Non-Max Suppression and vector generalisation to handle the high dimensionality of the output. The output of the YOLO algorithm is a tensor that contains the bounding box coordinates, ObjectNess codes and class probabilities. This high dimensional tensor is flattened into a vector to make it easier to process and then the vector is passed through the softmax function to convert class core probabilities
