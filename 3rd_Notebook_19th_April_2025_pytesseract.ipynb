```python
import pytesseract
from PIL import Image
import cv2
# PIL is the python imaging library which provides the python interpreter with image editing capabilities.
# This is so because python cannot directly edit the image. The Image module provides a class with the same name
# which is used to represent the PIL image.
```

```python
img=Image.open("img1.png")
text=pytesseract.image_to_string(img)
print(text)
```

```python
img=Image.open("img2.jpg")
text=pytesseract.image_to_string(img)
print(text)
```

```python
img1=cv2.imread("img1.png")
img1 # BGR
img1_rgb=cv2.cvtColor(img1,cv2.COLOR_BGR2RGB)
text=pytesseract.image_to_string(img1_rgb)
print(text)
```

```python
img2=cv2.imread("img2.jpg")
img2 # BGR
img2_rgb = cv2.cvtColor(img2,cv2.COLOR_BGR2RGB)
text = pytesseract.image_to_string(img2_rgb)
print(text)
```

```python
print(pytesseract.image_to_boxes(img1_rgb))
```

```python
df=pytesseract.image_to_data(img1_rgb,output_type="data.frame")
df
# the -1 means that the item is not a word. the high confidence results are generally
# the one's which have the conf of greater than 60
```

```python
img3=cv2.imread("invoice 1.jpg")
img3 # BGR
img3_rgb = cv2.cvtColor(img3,cv2.COLOR_BGR2RGB)
text = pytesseract.image_to_string(img3_rgb)
print(text)
```

```python
import re

re.search(r"[0-9]{2}/[0-9]{2}/[0-9]{4}",text).group()
```

```python
re.findall(r"[0-9]{2}/[0-9]{2}/[0-9]{4}",text)
```

```python
df=pytesseract.image_to_data(img3_rgb,output_type="data.frame")
df
```

```python
for i,t in enumerate(df["text"]):
  print(i,t)
```

```python
import os
invoices = os.listdir("INVOICES")
print(invoices)
```

```python
import pandas as pd
```

```python
prices=[]
dates=[]
for i in invoices:
  invoice=cv2.imread(f"INVOICES/{i}")
  invoice_rgb=cv2.cvtColor(invoice,cv2.COLOR_BGR2RGB)
  text=pytesseract.image_to_string(invoice_rgb)
  df=pytesseract.image_to_data(invoice_rgb,output_type="data.frame")
  dates.append(df["text"][43])
  prices.append(df["text"][119])
print(dates)
print(prices)
```

```python
import pandas as pd

df = pd.DataFrame({
    'Date': dates,
    'Price': prices
})
df
```
